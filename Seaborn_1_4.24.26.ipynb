```python
# import libraries
import pandas as pd

import matplotlib.pyplot as plt

import seaborn as sns
```

```python
# Load dataset
df = pd.read_excel("Kering Financial Data.xlsx")
```

```python
# preview data
df
```

```python
# Basic cleaning (optional): drop rows with missing values
df = df.dropna()
```

```python
# preview data to test if worked
df
```

```python
# Pandas visualization (replace the column names
df.plot(
    kind = 'bar',
    x = 'Brand',
    y = ['Revenue_2025_MEuro', 'Revenue_2024_MEuro'],
    figsize = (10,6)
)

plt.title('Revenue Comparison by Brand')
plt.xlabel('Brand')
plt.ylabel('Revnue (Million Euro)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Seaborn bar plot
long_df = df.melt(
    id_vars = 'Brand',
    value_vars = ['Revenue_2025_MEuro', 'Revenue_2024_MEuro'],
    var_name = 'Year',
    value_name = 'Revenue'
)

sns.barplot(
    data = long_df,
    x = 'Brand', 
    y = 'Revenue',
    hue = 'Year'
)

plt.title('Revenue Comparison by Brand (Seaborn)')
plt.xlabel('Brand')
plt.ylabel('Revenue (Million)Euro')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# Seaborn heatmap (correlation between the same revenue columns)
correlation = df[['Revenue_2025_MEuro', 'Revenue_2024_MEuro']].corr()

sns.heatmap(correlation, annot = True)

plt.title('Correlation Between 2024 and 2025 Revenue')
plt.tight_layout()
plt.show()
```

```python
# Seaborn distribution plot
sns.histplot(
    data = df,
    x = 'Revenue_2025_MEuro',
    kde = True
)

plt.title('Distribution of 2025 Revenue')
plt.xlabel('Revenue (Million Euro)')
plt.tight_layout()
plt.show
```

```python
# Line plot, revenue by brand over time
df = pd.read_excel("Kering Full Group Financials.xlsx")
```

```python
df
```

```python
brand_columns = [
    'Gucci_Revenue',
    'YSL_Revenue',
    'Bottega_Veneta_Revenue',
    'Other_Houses_Revenue',
    'Kering_Eyewear_Revenue',
]

brand_revenue_long = df.melt(
    id_vars = 'Year',
    value_vars = brand_columns,
    var_name = 'Brand',
    value_name = 'Revenue'
)

sns.lineplot(
    data = brand_revenue_long,
    x = 'Year',
    y = 'Revenue',
    hue = 'Brand',
    marker = 'o'

)

plt.title('Kering Brand Revenue Trends, 2014-2024')
plt.xlabel('Year')
plt.ylabel('Revenue (€ millions)')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
financials_long = df.melt(
    id_vars = 'Year',
    value_vars = ['Total_Revenue', 'Net_Profit'],
    var_name = 'Metric',
    value_name = 'Amount'
)

sns.lineplot(
    data = financials_long,
    x = 'Year',
    y = 'Amount',
    hue = 'Metric',
    marker = 'o'
)

plt.title('Kering Total Revenue vs Net Profit, 2014-2024')
plt.xlabel('Year')
plt.ylabel('Amount (€ millions)')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
revenue_2024 = df[df['Year'] == 2024].melt(
    id_vars = 'Year',
    value_vars = brand_columns,
    var_name = 'Brand',
    value_name = 'Revenue'
)

sns.barplot(
    data = revenue_2024,
    x = 'Brand',
    y = 'Revenue'
)

plt.title('Kering Revenue by Brand in 2024')
plt.xlabel('Brand')
plt.ylabel('Revenue (€ millions)')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
sns.stripplot(
    data = brand_revenue_long,
    x = 'Brand',
    y = 'Revenue',
    jitter = True
)

plt.title('Revenue Spread by Brand, 2014-2024')
plt.xlabel('Brand')
plt.ylabel('Revenue (€ millions)')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```
