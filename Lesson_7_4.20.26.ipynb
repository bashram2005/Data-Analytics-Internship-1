```python
# See all column names
df.columns
```

```python
# See all column names listed
df.columns.tolist()
```

```python
df['Social_average'] = df[['S1: Demographic Pressures', 'S2: Refugees and IDPs']].mean(axis=1)
df['Economic_average'] = df[['E1: Economy', 'E2: Economic Inequality', 'E3: Human Flight and Brain Drain']].mean(axis=1)
df['Political_average'] = df[['P1: State Legitimacy', 'P2: Public Services', 'P3: Human Rights']].mean(axis=1)
df['Cohesion_average'] = df[['C1: Security Apparatus', 'C2: Factionalized Elites', 'C3: Group Grievance']].mean(axis=1)
```

```python
import matplotlib.pyplot as plt

# Average of each category across all countries
category_means = df[['Social_average', 'Economic_average', 'Political_average', 'Cohesion_average']].mean()

category_means.plot(kind = 'bar', figsize =(10,6))

plt.title('Average Scores by Category')
plt.xlabel('Category')
plt.ylabel('Average Score')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
