```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        self.is_read = False
    def printBookinfo(self):
        print(self.title)
        print(self.author)
        print(self.pages)
        print(self.is_read)

book1 = Book('1984', 'George Orwell', 328)
book1.printBookinfo()
```
